# Bnet Inference

In this notebook we do inference with our drought trigger network. 

Specifically, we study how conditioning on 
* X1, X2, X3,
* FAR, HR, and AUROC,
* ProbImpactTail

changes the 
distribution of ProbBound.

```python
# this tells it to start looking for things from the top repo folder down.
import os
import sys
os.chdir('../')
sys.path.insert(0,os.getcwd())
print(os.getcwd())
```

```python
from bnet import *
```

```python
path = "data/leaf-nodes.csv"
ins_costs = [3, 4, 3]
rep_costs = [6, 12, 5]
bn = build_bnet(path, ins_costs, rep_costs)
```

```python
import pyagrum.lib.notebook as gnb
gnb.showBN(bn)
```

```python
gnb.showInference(bn,evs={},targets={},size="11")
```

```python
evs = {"FAR": 1, "HR": 2, "AUROC": 2, "X1": 0, "X2": 1, "X3": 1, "ProbImpactTail": 1}
gnb.showInference(bn,evs=evs,targets={},size="11")
```

```python
# change FAR evidence from 1 to 0, AUROC from 2 to 3
evs = {"FAR": 0, "HR": 3, "AUROC": 3, "X1": 0, "X2": 1, "X3": 1, "ProbImpactTail": 1}
gnb.showInference(bn,evs=evs,targets={},size="11")
```
