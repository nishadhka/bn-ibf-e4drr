# Bnet creation

In this notebook, we create a toy drought trigger bnet using pyagrum and
supporting code of our own

```python
# this makes sure it starts looking for things from the scumpy folder down.
import os
import sys
os.chdir('../')
sys.path.insert(0,os.getcwd())
print(os.getcwd())
```

```python
from bnet import *
```

```python
path = "data/leaf-nodes.csv"
ins_costs = [3, 4, 3]
rep_costs = [6, 12, 5]
bn = build_bnet(path, ins_costs, rep_costs)
```

```python
import pyagrum.lib.notebook as gnb
gnb.showBN(bn)
```

```python
for node_id in bn.nodes():
    node_name = bn.variable(node_id).name()
    print(f"CPT for {node_name}:")
    print(bn.cpt(node_id))
    gnb.sideBySide(bn.cpt(node_id))
    print("-" * 40)
```
